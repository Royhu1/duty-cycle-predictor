```python
from dcpredictor import driving_cycle_predictor
import pandas as pd
```

```python
dcpredictor = driving_cycle_predictor()
route_df = pd.read_csv('route.csv')
departure_time = pd.to_datetime('2025-04-12 08:00:00')
driving_cycle = dcpredictor.create_driving_cycle(route_df, departure_time)
print(driving_cycle.head())
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(10, 5))
plt.plot(driving_cycle['timestamp'], driving_cycle['speed'], label='Speed (m/s)')
plt.xlabel('Time')
plt.ylabel('Speed (m/s)')
plt.title('Driving Cycle Speed Profile')
plt.legend()
plt.grid()
plt.show()
```
